# ⭐ **10. Dictionaries in Python (Key–Value Data Storage)**

A **dictionary** stores data in **pairs**:

```
key → value
```

Think of it like a **real-world dictionary**:

* A word = **key**
* Its meaning = **value**

Python dictionary works the same way.

---

# 🔶 **1. Creating a Dictionary**

```python
student = {
    "name": "Vivek",
    "age": 21,
    "city": "Hyderabad"
}
```

✔ Keys are unique
✔ Values can be anything
✔ Fast to search
✔ Best for *real-world structured data*

---

# 🔶 **2. Accessing Dictionary Values**

```python
print(student["name"])   # Vivek
print(student["age"])    # 21
```

---

# 🔶 **3. Adding New Key–Value Pairs**

```python
student["grade"] = "A"
```

---

# 🔶 **4. Updating Existing Values**

```python
student["city"] = "Bangalore"
```

---

# 🔶 **5. Deleting Items (CRUD Operations)**

CRUD = Create, Read, Update, Delete
Dictionaries support all four.

---

## 🟥 Delete using `del`

```python
del student["age"]
```

---

## 🟥 Delete using pop()

```python
student.pop("name")
```

Returns the removed value.

---

## 🟥 Delete all items (clear)

```python
student.clear()
```

---

# ⭐ **6. Dictionary Methods (Very Important!)**

---

## 🔸 keys()

Returns all keys.

```python
print(student.keys())
```

---

## 🔸 values()

Returns all values.

```python
print(student.values())
```

---

## 🔸 items()

Returns both key & value pairs.

```python
print(student.items())
```

---

## 🔸 get()

Safe way to access values.

```python
print(student.get("name"))
print(student.get("country", "Not Found"))
```

Prevents errors if key doesn’t exist.

---

## 🔸 update()

Add or modify multiple pairs at once.

```python
student.update({"age": 22, "city": "Delhi"})
```

---

# ⭐ **7. Looping Through Dictionaries**

---

## 🔸 Loop through keys

```python
for key in student:
    print(key)
```

---

## 🔸 Loop through values

```python
for value in student.values():
    print(value)
```

---

## 🔸 Loop through key-value pairs

```python
for key, value in student.items():
    print(key, ":", value)
```

---

# ⭐ **8. Nested Dictionaries (Real-World Example)**

You can put one dictionary inside another.

```python
students = {
    1: {"name": "Vivek", "age": 21},
    2: {"name": "Arun", "age": 22}
}

print(students[1]["name"])  # Vivek
```

---

# ⭐ **9. Dictionary vs List — When to Use What?**

| Use a List when:        | Use a Dictionary when:  |
| ----------------------- | ----------------------- |
| Order matters           | Fast lookup is needed   |
| You store similar items | You store mapped data   |
| Items accessed by index | Items accessed by key   |
| Example: fruits list    | Example: student record |

# ⭐ Practical Mini Examples

---

## ✔ Example 1: Storing student info

```python
student = {
    "name": "Vivek",
    "age": 20,
    "course": "Python"
}
print(student["course"])
```

---

## ✔ Example 2: Phone contacts app

```python
contacts = {
    "Vivek": "9876543210",
    "Arun": "9123456780"
}

name = input("Whose number you want? ")

if name in contacts:
    print("Number:", contacts[name])
else:
    print("Contact not found.")
```

---

## ✔ Example 3: Word frequency counter (common interview task)

```python
text = "apple banana apple mango banana apple"
words = text.split()

freq = {}

for w in words:
    freq[w] = freq.get(w, 0) + 1

print(freq)
```

Output:

```
{'apple': 3, 'banana': 2, 'mango': 1}
```
