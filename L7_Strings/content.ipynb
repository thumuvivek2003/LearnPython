# ⭐ **7. Strings in Python (Text Handling for Beginners)**

A **string** is simply **text inside quotes**.

Examples:

```python
name = "Vivek"
city = 'Hyderabad'
message = "Hello Python!"
```

Strings are used for:

* storing names
* messages
* user input
* addresses
* any text

---

# 🔶 **1. String Basics**

### ✔ Strings can use **single** or **double quotes**

```python
a = "Hello"
b = 'World'
```

Both are same.

---

### ✔ Strings behave like a list of characters

Example:

```python
text = "python"
```

Positions (indexing) start from 0:

| p | y | t | h | o | n |
| - | - | - | - | - | - |
| 0 | 1 | 2 | 3 | 4 | 5 |

Accessing characters:

```python
text[0]   # p
text[3]   # h
```

---

# ⭐ **2. String Slicing (Extract Parts of a String)**

Syntax:

```python
string[start : end : step]
```

### ✔ Example 1: Basic slice

```python
s = "Python"
print(s[0:2])   # Py
print(s[2:6])   # thon
```

### ✔ Example 2: Omitting start or end

```python
print(s[:3])   # Pyt
print(s[3:])   # hon
```

### ✔ Example 3: Reverse string

```python
print(s[::-1])   # nohtyP
```

### ✔ Example 4: Skip characters

```python
print(s[0:6:2])   # Pto
```

Slicing = very powerful!

# ⭐ **3. Useful Built-in String Functions**

Python gives many built-in functions to handle text.

---

## 🔸 Changing case

```python
s = "hello World"

print(s.upper())   # HELLO WORLD
print(s.lower())   # hello world
print(s.title())   # Hello World
print(s.capitalize())  # Hello world
```

---

## 🔸 Checking contents

```python
print("python".isalpha())   # True (only letters)
print("1234".isdigit())     # True (only digits)
print("Hello123".isalnum()) # True (letters + numbers)
```

---

## 🔸 Remove spaces

```python
msg = "  hello  "
print(msg.strip())   # removes both sides
print(msg.lstrip())  # left only
print(msg.rstrip())  # right only
```

---

## 🔸 Replace text

```python
s = "I love apples"
print(s.replace("apples", "mangoes"))
```

Output:

```
I love mangoes
```

---

## 🔸 Split and join

### split() → breaks string into list

```python
name = "Vivek Thumu"
print(name.split())   # ['Vivek', 'Thumu']
```

### join() → join list into string

```python
words = ["Python", "is", "fun"]
print(" ".join(words))
```

Output:

```
Python is fun
```

---

## 🔸 Counting occurrences

```python
s = "banana"
print(s.count("a"))   # 3
```

---

## 🔸 Find position

```python
s = "hello"
print(s.find("l"))    # 2 (first l)
```

---

# ⭐ **4. String Formatting (VERY Important)**

Formatting is combining text + variables neatly.

---

## 🔷 **Method 1: f-strings (BEST & easiest)**

```python
name = "Vivek"
age = 21

print(f"My name is {name} and I am {age} years old.")
```

---

## 🔷 **Method 2: format()**

```python
print("My name is {} and I am {} years old".format(name, age))
```

---

## 🔷 **Method 3: + operator**

```python
print("Hello " + name)
```

---

## 🔷 **Method 4: comma in print**

```python
print("Hello", name)
```

# ⭐ Mini Practical Examples

---

## ✔ Example 1: Reverse user input

```python
text = input("Enter text: ")
print("Reversed:", text[::-1])
```

---

## ✔ Example 2: Count vowels

```python
text = input("Enter word: ")

vowels = 0
for ch in text.lower():
    if ch in "aeiou":
        vowels += 1

print("Vowels:", vowels)
```

---

## ✔ Example 3: Simple password check

```python
pwd = input("Enter password: ")

if len(pwd) < 6:
    print("Password too short!")
else:
    print("Password accepted.")
```
