# ⭐ **6. Loops in Python (for & while)**

Loops allow you to **repeat actions automatically** without writing code again and again.

### Without loops:

You would need to write:

```python
print(1)
print(2)
print(3)
print(4)
print(5)
```

### With loops:

```python
for i in range(1, 6):
    print(i)
```

Much easier, right?

---

# 🔶 Why Do We Use Loops?

Loops help you:

* repeat tasks
* process lists
* count things
* automate repetitive work
* run actions until a condition is met

---

# ⭐ **1. for Loop (Repeat for a fixed number of times)**

### Structure:

```python
for variable in sequence:
    repeat this block
```

The most common sequence is `range()`.

---

# 🔷 **Using range()**

`range(start, stop, step)`

Examples:

| Code              | Meaning               |
| ----------------- | --------------------- |
| `range(5)`        | 0 to 4                |
| `range(1, 6)`     | 1 to 5                |
| `range(1, 10, 2)` | odd numbers 1,3,5,7,9 |

---

# ⭐ Example 1: Print 1 to 5

```python
for i in range(1, 6):
    print(i)
```

---

# ⭐ Example 2: Print each character in a string

```python
for ch in "python":
    print(ch)
```

Output:

```
p
y
t
h
o
n
```

---

# ⭐ Example 3: Loop through a list

```python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
```

---

# ⭐ Example 4: Print even numbers

```python
for i in range(2, 11, 2):
    print(i)
```

---

# ⭐ **2. while Loop (Repeat UNTIL a condition becomes False)**

Syntax:

```python
while condition:
    repeat this
```

The loop continues as long as the condition is **True**.

---

# ⭐ Example 1: Print 1 to 5

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

### Importance:

❗ If you forget `i += 1`, the loop becomes infinite!

---

# ⭐ Example 2: Keep asking until user says "stop"

```python
command = ""

while command != "stop":
    command = input("Type something (or 'stop'): ")
```

---

# ⭐ Example 3: Countdown

```python
i = 5

while i > 0:
    print(i)
    i -= 1
```

Output:

```
5
4
3
2
1
```

---

# ⭐ for vs while — When to Use What?

| Use **for loop** when…             | Use **while loop** when…                   |
| ---------------------------------- | ------------------------------------------ |
| You know how many times to repeat  | You repeat until a condition becomes false |
| Iterating over list, string, range | Waiting for user input                     |
| Counting numbers                   | Running until a stop condition             |

# 🔥 ⭐ Mini Task Examples (Very Simple)

## ✔ Example: Print sum of numbers 1 to 10

```python
total = 0

for i in range(1, 11):
    total += i

print(total)
```

---

## ✔ Example: Ask user until they enter 10

```python
num = 0

while num != 10:
    num = int(input("Enter 10 to stop: "))

print("Thank you!")
```

# ⭐ Common Beginner Mistakes

### ❌ Mistake 1: Infinite Loop

```python
while i < 5:
    print(i)
    # forgot i += 1
```

### ❌ Mistake 2: Wrong indentation

```python
for i in range(5):
print(i)    # ❌ wrong
```

Always indent with **4 spaces**.
