# ⭐ **16. Working with Packages using pip

Install & Use Thousands of Ready-Made Python Tools**

Until now, you wrote Python code yourself.

But Python has **THOUSANDS** of ready-made tools created by experts that can save you HOURS or DAYS of work.

Examples:

* **numpy** → advanced maths
* **pandas** → data analysis
* **matplotlib** → charts
* **requests** → call APIs
* **flask / django** → web apps

To use these tools, we install them using **pip**.

---

# 🔶 **1. What is pip?**

`pip` stands for:

**P**ackage
**I**nstaller for
**P**ython

It is used to:

* install packages
* uninstall packages
* upgrade packages
* list installed packages

pip is like the **Play Store** for Python.

---

# ⭐ **2. How to Install a Package Using pip**

Open **terminal / command prompt** and type:

```bash
pip install package_name
```

Example:

```bash
pip install numpy
```

pip downloads and installs numpy automatically.

---

# ⭐ **3. How to Use the Installed Package**

After installation, import it in Python:

```python
import numpy as np

arr = np.array([1, 2, 3])
print(arr)
```

---

# ⭐ **4. Installing Multiple Packages**

```bash
pip install pandas matplotlib seaborn
```

---

# ⭐ **5. Upgrading a Package**

```bash
pip install --upgrade package_name
```

Example:

```bash
pip install --upgrade pandas
```

---

# ⭐ **6. Uninstalling a Package**

```bash
pip uninstall package_name
```

Example:

```bash
pip uninstall numpy
```

---

# ⭐ **7. Check Installed Packages**

```bash
pip list
```

Shows all installed packages and versions.

---

# ⭐ **8. Search for a Package**

```bash
pip search keyword
```

Example:

```bash
pip search weather
```

# ⭐ **10. Requirements File (Advanced Beginner)**

If you create a project with many packages, you save them in:

```
requirements.txt
```

Create it using:

```bash
pip freeze > requirements.txt
```

Install everything from the file:

```bash
pip install -r requirements.txt
```

This helps when sharing projects with others.

# ⭐ **11. Real-Life Practical Examples**

---

## ✔ Example 1: Fetch data from internet (using requests)

Install:

```bash
pip install requests
```

Use:

```python
import requests

response = requests.get("https://api.github.com")
print(response.status_code)
print(response.json())
```

---

## ✔ Example 2: Chart using matplotlib

Install:

```bash
pip install matplotlib
```

Use:

```python
import matplotlib.pyplot as plt

plt.plot([1,2,3], [4,5,6])
plt.show()
```

---

## ✔ Example 3: Create arrays using numpy

Install:

```bash
pip install numpy
```

Use:

```python
import numpy as np

arr = np.array([10, 20, 30])
print(arr * 2)
```
