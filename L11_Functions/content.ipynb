# ⭐ **11. Functions in Python (Write Reusable & Clean Code)**

A **function** is a reusable block of code that performs a specific task.

Think of a function like a **mini machine**:

* You give input
* It does some work
* It gives output

Instead of writing code again and again, put it inside a function and **reuse it anywhere**.

---

# 🔶 **1. Defining a Function (using `def`)**

Basic structure:

```python
def function_name():
    # code
```

Example:

```python
def greet():
    print("Hello from Python!")
```

Call the function:

```python
greet()
```

Output:

```
Hello from Python!
```

---

# 🔶 **2. Why Use Functions?**

✔ Avoid repeating code
✔ Code becomes clean
✔ Easy to update
✔ Helps in organizing big programs
✔ Reusable logic

Example: Instead of writing 10 lines again and again, put them in a function.

---

# ⭐ **3. Functions with Parameters (Inputs)**

Parameters allow functions to accept **values**.

Example:

```python
def greet(name):
    print("Hello", name)
```

Call it:

```python
greet("Vivek")
greet("Arun")
```

Output:

```
Hello Vivek
Hello Arun
```

Each time you pass different data.

---

# 🔶 Multiple Parameters

```python
def add(a, b):
    print(a + b)
```

Call:

```python
add(10, 20)
```

---

# ⭐ **4. return Statement (Output of a Function)**

`return` sends a value back to the place where the function was called.

Example:

```python
def add(a, b):
    return a + b
```

Use the returned value:

```python
result = add(5, 7)
print(result)
```

Output:

```
12
```

If you forget `return`, the function gives `None`.

---

# 🔶 Why return is important?

Because it allows functions to **produce results** instead of just printing.

---

# ⭐ **5. Function with Input + Output Example**

```python
def check_even(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even(10))   # Even
print(check_even(7))    # Odd
```

---

# ⭐ **6. Default Parameter Values**

You can give default values:

```python
def greet(name="Guest"):
    print("Hello", name)
```

Call:

```python
greet()          # Hello Guest
greet("Vivek")   # Hello Vivek
```

---

# ⭐ **7. Keyword Arguments**

Call by naming parameters:

```python
def intro(name, age):
    print(f"My name is {name} and I am {age} years old.")

intro(age=21, name="Vivek")
```

Order doesn’t matter now.

---

# ⭐ **8. Functions Returning Multiple Values**

Python can return multiple outputs as a tuple:

```python
def calculate(a, b):
    return a + b, a - b, a * b

add, sub, mul = calculate(10, 5)

print(add, sub, mul)
```

---

# ⭐ **9. Scope — Local vs Global Variables**

### Local variable

Defined **inside a function**. Only works inside it.

```python
def demo():
    x = 10   # local variable
    print(x)
```

### Global variable

Defined outside the function.

```python
x = 100

def show():
    print(x)

show()
```

## ✔ Example 1: Simple Calculator

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

print(add(10, 5))
print(subtract(10, 5))
```

---

## ✔ Example 2: Factorial using function

```python
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

print(factorial(5))   # 120
```

---

## ✔ Example 3: Count vowels in a string

```python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch in "aeiou":
            count += 1
    return count

print(count_vowels("Python"))
```

---

## ✔ Example 4: Check login

```python
def login(username, password):
    if username == "admin" and password == "123":
        return "Login success"
    return "Invalid credentials"

print(login("admin", "123"))
```
