# ⭐ **15. Introduction to OOP (Object-Oriented Programming)**

OOP is a way of writing programs by creating **objects**, like we have objects in the real world.

Think of real-world objects:

* Car
* Dog
* Student
* Bank Account

Each real object has:

✔ **Attributes** (data)
✔ **Methods** (actions)

Python lets us model things the same way using:

* **Class** → blueprint
* **Object** → actual item created from blueprint

---

# 🔶 **1. What is a Class? (Blueprint)**

A **class** is a design or blueprint.

Example:
A **"Car"** class describes what every car *should have*:

* Attributes: color, speed, brand
* Methods: start(), stop(), accelerate()

But a class is NOT a car — it’s only a plan.

---

# 🔶 **2. What is an Object? (Real thing)**

An **object** is created from the class.

Example:

```python
my_car = Car()
```

Now `my_car` is a real object with its own data.

---

# ⭐ **3. Creating a Simple Class (Beginner Friendly)**

```python
class Person:
    pass
```

This creates a class but with no data yet.

---

# ⭐ **4. Adding Attributes & Methods**

A **method** is just a function inside a class.

Let’s create a Person class:

```python
class Person:
    def __init__(self, name, age):
        self.name = name     # attribute
        self.age = age       # attribute

    def greet(self):         # method
        print(f"Hello, I am {self.name} and I am {self.age} years old.")
```

---

# 🔶 Explanation:

### ✔ `__init__()`

This is a **constructor** — runs automatically when you create an object.
It is used to give initial values to the object.

### ✔ `self`

Refers to the current object.
Every object has its own data, and `self` helps store it.

---

# ⭐ **5. Creating Objects from the Class**

```python
p1 = Person("Vivek", 21)
p2 = Person("Arun", 22)
```

---

# ⭐ **6. Using the Methods**

```python
p1.greet()
p2.greet()
```

Output:

```
Hello, I am Vivek and I am 21 years old.
Hello, I am Arun and I am 22 years old.
```

# 🔥 **7. Another Example: Car Class**

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def start(self):
        print(f"{self.brand} car is starting...")

    def stop(self):
        print(f"{self.brand} car is stopping...")
```

### Create objects:

```python
c1 = Car("BMW", "Black")
c2 = Car("Audi", "White")
```

### Call methods:

```python
c1.start()
c2.stop()
```

# ⭐ **9. Why Use OOP? (Simple Explanation)**

OOP helps you:

✔ Organize big programs easily
✔ Avoid repeating code
✔ Group data + functions
✔ Represent real-world things
✔ Make code reusable

Example:

Instead of writing separate variables for every student:

```
student1_name  
student1_age  
student1_roll  

student2_name  
student2_age  
student2_roll
```

You create a **Student class**, and create objects when needed.

# ⭐ **10. Mini Practical Examples**

---

## ✔ Example 1: Student Class

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def result(self):
        if self.marks >= 40:
            return "Pass"
        else:
            return "Fail"
```

Using it:

```python
s = Student("Vivek", 85)
print(s.result())
```

---

## ✔ Example 2: Bank Account Class

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal successful")
        else:
            print("Insufficient funds")
```

---

## ✔ Example 3: Dog Class

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        print(self.name, "is barking!")
```
