# ⭐ **3. Input & Output (I/O) in Python**

Think of your program as a small conversation:

* **Output:** Python **speaks** to you → using `print()`
* **Input:** Python **listens** to you → using `input()`

Let’s learn both in the simplest possible way.

---

# 🔶 **1. OUTPUT — Using `print()`**

`print()` is how Python shows information on the screen.

Example:

```python
print("Hello!")
```

Output:

```
Hello!
```

You can print:

* Text
* Variables
* Numbers
* Multiple values

Example:

```python
name = "Vivek"
age = 21
print("My name is", name, "and I am", age, "years old.")
```

Output:

```
My name is Vivek and I am 21 years old.
```

---

# 🔶 **2. INPUT — Using `input()`**

`input()` allows the user to type something.

It **always returns text (string)**.

Example:

```python
name = input("Enter your name: ")
print("Hello", name)
```

💡 Whatever you type becomes the value of `name`.

If user types:

```
Vivek
```

Output:

```
Hello Vivek
```

---

# ⭐ **Important: input() ALWAYS gives a string!**

If you want a number, you must convert it.

### Convert to integer:

```python
age = int(input("Enter your age: "))
print("Your age is", age)
```

### Convert to decimal:

```python
price = float(input("Enter price: "))
print("Price is", price)
```

---

# 🔶 **3. Combining Input + Output — Full Example**

### ⭐ Example 1: Greeting program

```python
name = input("What is your name? ")

print("Nice to meet you,", name)
```

---

### ⭐ Example 2: Add two numbers

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

total = num1 + num2
print("The total is:", total)
```

---

# 🔶 **4. Formatting Output (Making Print Beautiful)**

### 👉 Method 1: f-strings (BEST & simple)

```python
name = "Vivek"
age = 21

print(f"My name is {name} and I am {age} years old.")
```

Output:

```
My name is Vivek and I am 21 years old.
```

### 👉 Method 2: + operator

```python
print("Hello " + name)
```

---

# 🔶 **5. Real-Life Mini Project Example**

### ⭐ Program: Ask user for details and show summary

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
city = input("Enter your city: ")

print("\n--- User Profile ---")
print(f"Name : {name}")
print(f"Age  : {age}")
print(f"City : {city}")
```

Output example:

```
--- User Profile ---
Name : Vivek
Age  : 21
City : Hyderabad
```

Perfect way to practice **input + output** together.
