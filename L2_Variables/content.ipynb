# ⭐ 2. **Variables & Data Types (For Absolute Beginners)**

To write any program, you must store information — **this is where variables come in.**

---

# 🔶 **What is a Variable? (Ultra Simple Explanation)**

A variable is **a container that stores information**.

Imagine a **labeled box**:

* The **label** = variable name
* The **stuff inside** = data (value)

Example:

```python
age = 20
```

Here:

* `age` → name of the box
* `20` → data stored inside

You can change the data anytime:

```python
age = 25     # new value replaces old one
```

---

# 🔶 **Why are Variables Needed?**

Because programs need to **remember things** like:

* user’s age
* a score in a game
* a price
* a name

Variables help computers **store and reuse values**.

---

# ⭐ **Python Data Types You Must Know (4 Main Types)**

## 1️⃣ **int → Integer (Whole numbers)**

Used for **counting** and **math**.

Examples:

```python
age = 21
points = 150
year = 2025
```

---

## 2️⃣ **float → Decimal numbers**

Used for **precision numbers**, like height, weight, price.

Examples:

```python
pi = 3.14
height = 5.9
temperature = 36.6
```

---

## 3️⃣ **str → String (Text)**

Text inside **quotes** `" "` or `' '`.

Examples:

```python
name = "Vivek"
city = "Hyderabad"
message = "Welcome to Python!"
```

Python treats anything inside quotes as **text**, even numbers:

```python
x = "100"   # this is a string, not a number
```

---

## 4️⃣ **bool → Boolean (True or False)**

Used for decisions (conditions).

Examples:

```python
is_student = True
is_admin = False
logged_in = True
```

Booleans are always:

* **True**
* **False**

(First letter capital is important!)

---

# ⭐ **Putting It All Together**

```python
name = "Vivek"          # str
age = 21                # int
height = 5.8            # float
is_student = True       # bool

print(name)
print(age)
print(height)
print(is_student)
```

### Output:

```
Vivek
21
5.8
True
```

---

# 🔶 **How to Check a Variable’s Data Type**

Use `type()` function.

Example:

```python
x = 10
print(type(x))
```

Output:

```
<class 'int'>
```

Another example:

```python
y = "Hello"
print(type(y))
```

Output:

```
<class 'str'>
```

---

# 🔶 **Changing Values (Variables are flexible!)**

```python
score = 10
score = 20
score = score + 5
print(score)
```

Output:

```
25
```

# 🔶 **Common Beginner Mistakes**

### ❌ Mistake 1:

Using space in variable names

```python
my name = "Vivek"   # wrong
```

### ✔ Correct:

```python
my_name = "Vivek"
```

---

### ❌ Mistake 2:

```python
1age = 20   # wrong
```

### ✔ Correct:

```python
age1 = 20
```
