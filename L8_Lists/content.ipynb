# ⭐ **8. Lists in Python (Store Multiple Values Easily)**

A **list** is a container that stores **multiple items** in **one variable**.

Example:

```python
fruits = ["apple", "banana", "mango"]
```

Lists are:

* ordered
* changeable
* allow duplicates
* can store different types

Lists make handling large sets of data easy.

---

# 🔶 **1. Creating a List**

```python
numbers = [10, 20, 30, 40]
names = ["Vivek", "Arun", "Sai"]
mixed = [10, "hello", 3.14, True]
```

You can put ANY data inside a list.

---

# 🔶 **2. Accessing List Items (Indexing)**

Index starts from **0**:

| apple | banana | mango |
| ----- | ------ | ----- |
| 0     | 1      | 2     |

Example:

```python
fruits = ["apple", "banana", "mango"]

print(fruits[0])   # apple
print(fruits[2])   # mango
```

---

# 🔶 **3. Negative Indexing**

Starts from the **end**:

| apple | banana | mango |
| ----- | ------ | ----- |
| -3    | -2     | -1    |

Example:

```python
print(fruits[-1])   # mango
print(fruits[-2])   # banana
```

---

# 🔶 **4. Slicing Lists**

Same as strings:

```python
fruits = ["apple", "banana", "mango", "orange", "grapes"]

print(fruits[1:4])   # ['banana', 'mango', 'orange']
print(fruits[:3])    # ['apple', 'banana', 'mango']
print(fruits[2:])    # ['mango', 'orange', 'grapes']
```

---

# ⭐ **5. List Methods (Super Important!)**

Python provides many built-in methods to modify lists.

---

## 🔸 append() — Add at END

```python
fruits.append("pineapple")
```

---

## 🔸 insert() — Add at specific position

```python
fruits.insert(1, "kiwi")   # add at index 1
```

---

## 🔸 remove() — Remove a value

```python
fruits.remove("banana")
```

---

## 🔸 pop() — Remove by index (default = last)

```python
fruits.pop(2)  # removes item at index 2
fruits.pop()   # removes last item
```

---

## 🔸 sort() — Sort ascending

```python
numbers = [3, 1, 4, 2]
numbers.sort()
print(numbers)   # [1,2,3,4]
```

---

## 🔸 reverse()

```python
numbers.reverse()
```

---

## 🔸 count() — Count occurrences

```python
nums = [1,2,2,3]
print(nums.count(2))   # 2
```

---

## 🔸 index() — Find position

```python
names = ["a", "b", "c"]
print(names.index("b"))   # 1
```

---

## 🔸 extend() — Add multiple items

```python
a = [1,2]
b = [3,4]
a.extend(b)
print(a)   # [1,2,3,4]
```

---

# ⭐ **6. Change Values in List (Lists Are Mutable!)**

Unlike strings, lists **can be modified**.

```python
fruits = ["apple", "banana", "mango"]
fruits[1] = "kiwi"

print(fruits)  # ['apple', 'kiwi', 'mango']
```

---

# ⭐ **7. Looping Through Lists**

### 👉 for loop:

```python
for fruit in fruits:
    print(fruit)
```

---

### 👉 Using range():

```python
for i in range(len(fruits)):
    print(i, fruits[i])
```

---

# ⭐ **8. Check if an item exists**

```python
if "apple" in fruits:
    print("Apple found!")
```

---

# ⭐ **9. List Length**

```python
print(len(fruits))
```
