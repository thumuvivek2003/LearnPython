# 🔥 **PROJECT 1: Number Guessing Game**

(Level: Very Easy — uses input, loops, conditions, random)

You guess a number chosen by the computer.

```python
import random

secret = random.randint(1, 10)
guess = 0

while guess != secret:
    guess = int(input("Guess a number between 1 and 10: "))

    if guess < secret:
        print("Too low!")
    elif guess > secret:
        print("Too high!")
    else:
        print("Correct! You guessed the number!")
```

Concepts used:
✔ random
✔ int + input
✔ while loop
✔ if / elif / else

# 🔥 **PROJECT 2: To-Do List App**

(Level: Easy — uses lists, loops, files)

```python
tasks = []

while True:
    print("\n1. Add Task\n2. View Tasks\n3. Save & Exit")
    choice = input("Choose option: ")

    if choice == "1":
        task = input("Enter task: ")
        tasks.append(task)
    
    elif choice == "2":
        print("\nYour Tasks:")
        for t in tasks:
            print("-", t)
    
    elif choice == "3":
        with open("tasks.txt", "w") as f:
            for t in tasks:
                f.write(t + "\n")
        print("Tasks saved! Goodbye.")
        break
```

Concepts used:
✔ lists
✔ append
✔ file write
✔ loops + conditions

# 🔥 **PROJECT 3: Simple Calculator**

(Level: Easy — uses functions & input)

```python
def add(a, b): return a + b
def sub(a, b): return a - b
def mul(a, b): return a * b
def div(a, b): 
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"

while True:
    print("\n1.Add  2.Subtract  3.Multiply  4.Divide  5.Exit")
    choice = input("Choose: ")

    if choice == "5":
        break

    x = float(input("Enter first number: "))
    y = float(input("Enter second number: "))

    if choice == "1": print(add(x, y))
    elif choice == "2": print(sub(x, y))
    elif choice == "3": print(mul(x, y))
    elif choice == "4": print(div(x, y))
    else: print("Invalid choice")
```

Concepts used:
✔ input processing
✔ functions
✔ error handling
✔ loops & conditions

# 🔥 **PROJECT 4: Password Strength Checker**

(Level: Beginner — uses loops, conditions, string methods)

```python
pwd = input("Enter password: ")

strength = 0

if len(pwd) >= 8:
    strength += 1
if any(char.isdigit() for char in pwd):
    strength += 1
if any(char.isupper() for char in pwd):
    strength += 1
if any(char in "!@#$%^&*()" for char in pwd):
    strength += 1

if strength == 4:
    print("Strong Password 💪")
elif strength == 2 or strength == 3:
    print("Medium Password 🙂")
else:
    print("Weak Password 😢")
```

Concepts used:
✔ strings
✔ loops
✔ conditions

# 🔥 **PROJECT 5: Contact Book (Dictionary Project)**

(Level: Beginner — uses dictionaries, functions, loops, file handling)

```python
contacts = {}

while True:
    print("\n1.Add Contact 2.Search Contact 3.Exit")
    choice = input("Choose: ")

    if choice == "1":
        name = input("Name: ")
        number = input("Phone: ")
        contacts[name] = number

    elif choice == "2":
        name = input("Search name: ")
        print(contacts.get(name, "Not found"))

    elif choice == "3":
        with open("contacts.txt", "w") as f:
            for name, number in contacts.items():
                f.write(f"{name}, {number}\n")
        print("Saved & exiting...")
        break
```

Concepts used:
✔ dictionary
✔ get()
✔ loop
✔ file write

# 🔥 **PROJECT 6: Rock–Paper–Scissors Game**

(Level: Fun — uses random + logic)

```python
import random

options = ["rock", "paper", "scissors"]

user = input("Choose rock/paper/scissors: ").lower()
comp = random.choice(options)

print("Computer chose:", comp)

if user == comp:
    print("It's a tie!")
elif (user == "rock" and comp == "scissors") or \
     (user == "paper" and comp == "rock") or \
     (user == "scissors" and comp == "paper"):
    print("You win!")
else:
    print("Computer wins!")
```

Concepts used:
✔ list
✔ random
✔ condition checks

# 🔥 **PROJECT 7: Mini Bank System (OOP Example)**

(Level: Intermediate beginner — uses classes)

```python
class BankAccount:
    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
    
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return "Withdrawal successful"
        return "Insufficient balance"
    
    def details(self):
        return f"{self.name} - Balance: {self.balance}"

account = BankAccount("Vivek", 500)

print(account.details())
account.deposit(200)
print(account.details())
print(account.withdraw(100))
print(account.details())
```

Concepts used:
✔ OOP
✔ attributes
✔ methods
✔ logic
