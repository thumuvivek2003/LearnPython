# ⭐ **13. File Handling in Python (Read & Write Files Easily)**

Until now, your programs could only store data **temporarily** (in variables).
But what if you want to store data **permanently**?

✔ Notes
✔ User records
✔ Logs
✔ Text data
✔ Scores

For that, Python uses **files**.

---

# 🔶 **1. Opening a File (VERY IMPORTANT)**

Python uses the `open()` function:

```python
open("filename", "mode")
```

## Modes:

| Mode   | Meaning                       |
| ------ | ----------------------------- |
| `"r"`  | read                          |
| `"w"`  | write (overwrites file)       |
| `"a"`  | append (add new data to end)  |
| `"x"`  | create file (error if exists) |
| `"r+"` | read + write                  |

---

# ⭐ **2. Writing to a File**

### Step 1: open file in write mode

### Step 2: write text

### Step 3: close the file

Example:

```python
file = open("data.txt", "w")
file.write("Hello, this is Python file handling!")
file.close()
```

This creates (or overwrites) **data.txt**.

---

# ⭐ **3. Appending to a File (Don’t overwrite)**

```python
file = open("data.txt", "a")
file.write("\nAdding new line to the file.")
file.close()
```

This adds text without deleting old content.

---

# ⭐ **4. Reading a File (Super Easy)**

```python
file = open("data.txt", "r")
content = file.read()
print(content)
file.close()
```

---

# 🔶 **Different Read Methods**

### ✔ read()

Reads entire file.

```python
file.read()
```

---

### ✔ readline()

Reads **one line** at a time.

```python
line = file.readline()
```

---

### ✔ readlines()

Reads **all lines into a list**.

```python
lines = file.readlines()
print(lines)
```

---

# ⭐ **5. Best Practice: Using with (NO NEED TO CLOSE)**

`with` automatically closes the file for you.

### ⭐ Writing with `with`

```python
with open("data.txt", "w") as f:
    f.write("Hello World!")
```

### ⭐ Reading with `with`

```python
with open("data.txt", "r") as f:
    print(f.read())
```

This is the recommended method.

---

# ⭐ **6. File Handling with Error Handling**

```python
try:
    with open("info.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found!")
```

Prevents crashing if file missing.

---

# ⭐ **7. Writing User Input to File**

```python
with open("users.txt", "a") as f:
    name = input("Enter your name: ")
    f.write(name + "\n")
```

This stores user input permanently.

# ⭐ **8. Mini Project Examples**

---

## ✔ Example 1: Store and read notes

### Writing:

```python
with open("notes.txt", "w") as f:
    f.write("My first note in Python!")
```

### Reading:

```python
with open("notes.txt", "r") as f:
    print(f.read())
```

---

## ✔ Example 2: Count lines in a file

```python
with open("notes.txt", "r") as f:
    lines = f.readlines()
    print("Total lines:", len(lines))
```

---

## ✔ Example 3: Search word in a file

```python
word = input("Enter word to search: ")

with open("notes.txt", "r") as f:
    content = f.read()

if word in content:
    print("Word found!")
else:
    print("Not found.")
```

---

## ✔ Example 4: Save scores

```python
with open("scores.txt", "a") as f:
    name = input("Name: ")
    score = input("Score: ")
    f.write(f"{name} - {score}\n")

print("Score saved!")
```
