# ⭐ **4. Operators & Expressions (Beginner Friendly)**

Operators are **symbols that help Python do actions** like:

* add numbers
* compare values
* check conditions
* assign values

Think of operators as **math and decision tools** inside Python.

---

# 🔶 **1. Arithmetic Operators (Math Operators)**

Used for basic calculations.

| Operator | Meaning                     | Example   | Result |
| -------- | --------------------------- | --------- | ------ |
| `+`      | Addition                    | `5 + 3`   | 8      |
| `-`      | Subtraction                 | `10 - 4`  | 6      |
| `*`      | Multiplication              | `6 * 7`   | 42     |
| `/`      | Division                    | `10 / 3`  | 3.333  |
| `//`     | Floor Division (no decimal) | `10 // 3` | 3      |
| `%`      | Modulus (remainder)         | `10 % 3`  | 1      |
| `**`     | Power                       | `2 ** 3`  | 8      |

### Example:

```python
x = 10
y = 3

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x // y)
print(x % y)
print(x ** y)
```

---

# 🔶 **2. Comparison Operators (Checking Conditions)**

These return **True or False**.

| Operator | Meaning          | Example  | Result |
| -------- | ---------------- | -------- | ------ |
| `==`     | Equal to         | `5 == 5` | True   |
| `!=`     | Not equal        | `5 != 3` | True   |
| `>`      | Greater than     | `10 > 2` | True   |
| `<`      | Less than        | `3 < 7`  | True   |
| `>=`     | Greater or equal | `5 >= 5` | True   |
| `<=`     | Less or equal    | `3 <= 5` | True   |

### Example:

```python
age = 18
print(age >= 18)   # True
print(age < 18)    # False
```

---

# 🔶 **3. Logical Operators (Decision Making)**

Used when you combine multiple conditions.

| Operator | Meaning                 | Example                   |
| -------- | ----------------------- | ------------------------- |
| `and`    | True if BOTH are true   | `age >= 18 and age <= 60` |
| `or`     | True if ANY one is true | `age < 18 or age > 60`    |
| `not`    | Reverses True/False     | `not(age == 18)`          |

### Example:

```python
age = 20

print(age > 18 and age < 30)  # True because both conditions true
print(age < 18 or age > 30)   # False because none are true
print(not(age == 20))         # False (reverse)
```

Logical operators help Python **make decisions**.

---

# 🔶 **4. Assignment Operators**

Used to **store values** in variables.

| Operator | Meaning           | Example  | Same As   |
| -------- | ----------------- | -------- | --------- |
| `=`      | Assign            | `x = 5`  | x = 5     |
| `+=`     | Add & assign      | `x += 3` | x = x + 3 |
| `-=`     | Subtract & assign | `x -= 2` | x = x - 2 |
| `*=`     | Multiply & assign | `x *= 4` | x = x * 4 |
| `/=`     | Divide & assign   | `x /= 5` | x = x / 5 |

### Example:

```python
x = 10
x += 5    # x = x + 5
print(x)  # 15

x -= 3
print(x)  # 12
```

Assignment operators help you update values easily.

# 🔶 **Putting It All Together: Small Mini Examples**

### ⭐ Example 1: Check if you can vote

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("You can vote!")
else:
    print("You cannot vote.")
```

Uses:
✔ Comparison (`>=`)
✔ Input
✔ Output

---

### ⭐ Example 2: Check if a number is even

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

Uses:
✔ Arithmetic `%`
✔ Comparison `==`
✔ Input-output

---

### ⭐ Example 3: Salary increment using assignment operator

```python
salary = 20000
salary += 5000
print("New salary =", salary)
```
