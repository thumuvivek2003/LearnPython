# ⭐ **12. Error Handling in Python (try–except)**

Errors happen **all the time** in programming.

Example errors:

* Dividing by zero
* Using undefined variables
* Wrong data input
* Opening missing files

Instead of letting the program **crash**, Python allows you to **catch the error** and handle it safely.

That’s what **exceptions** and **try–except** are for.

---

# 🔶 **1. What is an Exception? (Simple Explanation)**

An **exception** = a runtime error.

Example:

```python
print(10 / 0)
```

Output:

```
ZeroDivisionError: division by zero
```

This **stops the program** unless you handle it.

---

# ⭐ **2. Basic try-except (Beginner Level)**

Structure:

```python
try:
    # code that may cause error
except:
    # what to do if error happens
```

Example:

```python
try:
    num = int(input("Enter a number: "))
    print("You entered:", num)
except:
    print("Invalid input! Please enter a number.")
```

If user types "abc", this prevents the crash.

---

# ⭐ **3. Catching Specific Errors (Professional Way)**

You can catch **specific** errors using:

```python
except ValueError:
except ZeroDivisionError:
except TypeError:
```

Example:

```python
try:
    x = int(input("Enter number: "))
    print(10 / x)
except ValueError:
    print("Please enter a valid number!")
except ZeroDivisionError:
    print("You cannot divide by zero!")
```

---

# ⭐ **4. Using else and finally (Optional But Important)**

### ✔ **else** runs **only if no error occurs**

```python
try:
    x = int(input("Enter number: "))
except:
    print("Error!")
else:
    print("No error! You entered:", x)
```

---

### ✔ **finally** always runs (error or no error)

Good for:

* closing files
* releasing resources
* cleaning up

```python
try:
    print("Trying something...")
except:
    print("Error happened.")
finally:
    print("This runs no matter what.")
```

---

# ⭐ **5. Raising Your Own Errors (beginner-friendly)**

Use `raise` to force an error intentionally.

Example:

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative!")
```

Useful when validating inputs.

---

# ⭐ **6. Real-Life Examples (Very Easy)**

---

## ✔ Example 1: Safe division

```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero."
```

Call:

```python
print(divide(10, 0))
```

---

## ✔ Example 2: Safe user input

```python
while True:
    try:
        age = int(input("Enter your age: "))
        break
    except ValueError:
        print("Please enter numbers only!")
```

Loop continues until valid input.

---

## ✔ Example 3: File handling

```python
try:
    f = open("data.txt")
    print(f.read())
except FileNotFoundError:
    print("File not found!")
```

---

## ✔ Example 4: Multiple exceptions

```python
try:
    x = int(input("Enter number: "))
    y = 10 / x
except ValueError:
    print("Invalid number!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
```
