# 🔶 **1. What is a Module?**

A **module** is just a Python file containing ready-made functions, variables, or classes.

Examples of built-in modules:

* `math`
* `random`
* `datetime`
* `os`

Using these saves **time and effort**.

---

# 🔶 **2. How to Import a Module?**

### ✔ Method 1: Import whole module

```python
import math
print(math.sqrt(25))
```

---

### ✔ Method 2: Import specific function

```python
from math import sqrt
print(sqrt(25))
```

---

### ✔ Method 3: Import with alias (nickname)

```python
import math as m
print(m.pi)
```

---

# ⭐ **3. Popular Built-in Modules (Beginner Must Know)**

---

# 🔷 **math Module** (Mathematical functions)

```python
import math

print(math.sqrt(16))     # 4.0
print(math.pow(2, 3))    # 8
print(math.pi)           # 3.14159...
```

---

# 🔷 **random Module** (Random numbers)

```python
import random

print(random.randint(1, 10))     # Random number between 1 & 10
print(random.choice(["apple", "banana", "mango"]))
```

---

# 🔷 **datetime Module** (Dates & time)

```python
from datetime import datetime
now = datetime.now()
print(now)
```

---

# 🔷 **os Module** (Operating system functions)

```python
import os
print(os.listdir())   # show files in current folder
```

# ⭐ **6. Writing Your Own Module (Super Simple)**

Create a file **mymath.py**:

```python
def add(a, b):
    return a + b
```

Use it in another file:

```python
import mymath

print(mymath.add(10, 20))
```

You just created your **own module**!

# ⭐ **7. Practical Mini Examples**

---

## ✔ Example 1: Dice roll simulator

```python
import random

print("Dice rolled:", random.randint(1, 6))
```

---

## ✔ Example 2: Today's date

```python
from datetime import date

print("Today is:", date.today())
```

---

## ✔ Example 3: Generate 5 random OTP digits

```python
import random

otp = ""
for i in range(5):
    otp += str(random.randint(0,9))

print("Your OTP is:", otp)
```

---

## ✔ Example 4: Simple calculator using math

```python
import math
print(math.sin(0))
print(math.factorial(5))
```
