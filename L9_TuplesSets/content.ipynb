# ⭐ **9. Tuples & Sets in Python (Easy Explanation)**

Both **Tuples** and **Sets** are data structures like lists, but they have special purposes.

Let’s learn them in the simplest way.

---

# 🔶 **1. TUPLES — Immutable Lists (Fixed Data)**

A **tuple** is like a list, but **cannot be changed** once created.

Think of it like a **locked list**.

---

## ⭐ How to Create a Tuple

```python
numbers = (10, 20, 30)
student = ("Vivek", 21, "Hyderabad")
```

Notice:

* Uses **() parentheses**
* Can store different data types
* Ordered (index works)

---

## ⭐ Why Use Tuples?

Use tuples when the data **should NOT change**, such as:

* Coordinates (x, y)
* Dates
* Constants
* Configurations

They are:
✔ Faster than lists
✔ Safer (no accidental changes)

---

## ⭐ Accessing Tuple Items

Same as list indexing:

```python
numbers = (10, 20, 30)

print(numbers[0])  # 10
print(numbers[-1]) # 30
```

---

## ⭐ Tuples Are Immutable

This is **NOT allowed**:

```python
numbers[1] = 50   # ❌ ERROR
```

You **cannot modify**, **add**, or **remove** items in a tuple.

---

## ⭐ Useful Tuple Methods

### count()

```python
t = (1,2,2,3)
print(t.count(2))  # 2
```

### index()

```python
t = (10,20,30)
print(t.index(20))  # 1
```

---

# 🔥 Mini Example: Packing & Unpacking

```python
person = ("Vivek", 21, "India")

name, age, country = person

print(name)
print(age)
print(country)
```

---

# ⭐ **2. SETS — Unique & Unordered Data**

A **set** stores:

* only **unique** items
* in **no particular order**

Use sets when you want to remove duplicates or check membership fast.

---

## ⭐ How to Create a Set

```python
numbers = {1, 2, 3}
```

If duplicates exist, Python removes them:

```python
nums = {1,2,2,3,3}
print(nums)   # {1,2,3}
```

---

## ⭐ Set Properties

✔ Items are **unique**
✔ Items are **unordered** (no indexing)
✔ Very **fast** for membership checking
✔ Mutable (you can add/remove items)

---

# ⭐ Add Items to a Set

```python
s = {1, 2, 3}
s.add(4)
print(s)
```

---

# ⭐ Remove Items

```python
s.remove(2)   # remove item 2
```

If item doesn’t exist → ERROR.
Use `discard()` to avoid error:

```python
s.discard(100)  # no error even if 100 not present
```

---

# ⭐ Set Operations (VERY IMPORTANT)

Sets are amazing for **math-like operations**.

---

## 🔸 Union (combine sets)

```python
a = {1,2,3}
b = {3,4,5}

print(a | b)     # {1,2,3,4,5}
```

---

## 🔸 Intersection (common items)

```python
print(a & b)     # {3}
```

---

## 🔸 Difference (items in A not in B)

```python
print(a - b)     # {1,2}
```

---

## 🔸 Symmetric Difference (items not common)

```python
print(a ^ b)     # {1,2,4,5}
```

---

# ⭐ Convert List → Set (remove duplicates fast)

```python
nums = [1,2,2,3,3,4]
unique = set(nums)
print(unique)   # {1,2,3,4}
```

# ⭐ Practical Mini Examples

---

## ✔ Example 1: Remove duplicates from user input

```python
data = input("Enter numbers separated by space: ").split()
unique = set(data)
print("Unique items:", unique)
```

---

## ✔ Example 2: Check if item exists fast

```python
fruits = {"apple", "banana", "mango"}

if "apple" in fruits:
    print("Yes, apple is present!")
```

---

## ✔ Example 3: Tuple for fixed configuration

```python
color = (255, 0, 0)   # RGB values, fixed
print(color)
```

---

## ✔ Example 4: Set for operations

```python
a = {1,2,3,4}
b = {3,4,5,6}

print("Union:", a | b)
print("Intersection:", a & b)
```
