# ⭐ **5. Conditional Statements (if, elif, else)**

Conditional statements allow your program to **react** depending on the situation.

Just like humans make decisions:

* If raining → take umbrella
* If hungry → eat food
* Else → continue work

Python also needs this logic.
That’s what **if, elif, else** do.

---

# 🔶 **1. The `if` Statement (Python’s first decision)**

Syntax:

```python
if condition:
    # do something
```

Example:

```python
age = 20

if age >= 18:
    print("You are an adult.")
```

### 🧠 Meaning:

"If the condition is True, run the code inside."

Output:

```
You are an adult.
```

---

# 🔶 **2. The `else` Statement**

Used when the **if is False**.

```python
age = 15

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

Output:

```
Minor
```

Else means:
“If the above condition didn’t happen, do this.”

---

# 🔶 **3. The `elif` Statement (else + if)**

Used when you want **multiple conditions**.

Syntax:

```python
if condition1:
    ...
elif condition2:
    ...
else:
    ...
```

### Example: Grading system

```python
marks = 85

if marks >= 90:
    print("A Grade")
elif marks >= 75:
    print("B Grade")
elif marks >= 50:
    print("C Grade")
else:
    print("Fail")
```

Output:

```
B Grade
```

---

# ⭐ **4. Important Rules for Writing if–elif–else**

### ✔ Indentation is IMPORTANT

Python uses **spaces** to know what belongs to the block.

Correct:

```python
if age > 18:
    print("Adult")
```

Wrong:

```python
if age > 18:
print("Adult")   # ❌ No indentation, error
```

Always use **4 spaces** after `:`.

---

### ✔ Conditions must return True/False

Examples of valid conditions:

```python
age > 18
marks == 100
num % 2 == 0
```

---

### ✔ Only one `else` allowed

But many `elif` can be used.

# 🔶 **5. Real-Life Examples (Super Simple)**

---

## ⭐ Example 1: Check if a number is positive or negative

```python
num = int(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

---

## ⭐ Example 2: Check login access

```python
username = input("Enter username: ")

if username == "admin":
    print("Welcome Admin!")
else:
    print("Access Denied")
```

---

## ⭐ Example 3: Movie ticket discount

```python
age = int(input("Enter your age: "))

if age < 12:
    print("Child Discount")
elif age >= 60:
    print("Senior Citizen Discount")
else:
    print("No Discount")
```

---

## ⭐ Example 4: Even or Odd

```python
num = int(input("Enter number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

# 🔥 ⭐ Mini Project: ATM Simulation (Very Beginner Friendly)

```python
balance = 1000

choice = int(input("1. Check Balance\n2. Withdraw\nEnter choice: "))

if choice == 1:
    print("Your balance is", balance)
elif choice == 2:
    amount = int(input("Enter amount: "))
    if amount <= balance:
        print("Withdrawal successful!")
    else:
        print("Insufficient funds")
else:
    print("Invalid choice")
```

This combines:
✔ if
✔ elif
✔ else
✔ nested if
